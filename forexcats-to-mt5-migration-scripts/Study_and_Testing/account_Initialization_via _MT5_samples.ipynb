```python
import MetaTrader5 as mt5
import time

trading_acc_list = list(range(303000, 303061))
user_master_pass = "qwerty123"  # Мастер-пароль для всех аккаунтов
server = "123.45.67.890:443"  # Сервер

# Инициализация MetaTrader 5
if not mt5.initialize():
    print("Ошибка инициализации:", mt5.last_error())
    quit()

for login in trading_acc_list:
    # Авторизация на каждом аккаунте
    if mt5.login(login, user_master_pass, server):
        print(f"Успешная авторизация для счета: {login}")
        
        # Получение информации о счете
        account_info = mt5.account_info()
        if account_info is not None:
            print(f"Информация о счете {login}:")
            for field, value in account_info._asdict().items():
                print(f"{field}: {value}")
        else:
            print(f"Ошибка получения информации о счете {login}: {mt5.last_error()}")
    else:
        print(f"Ошибка авторизации для счета {login}: {mt5.last_error()}")

    time.sleep(5)  # Задержка между проверками

# Завершение соединения
mt5.shutdown()
```
