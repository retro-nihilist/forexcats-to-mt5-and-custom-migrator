```python
# параме запроса к базе торговых инструментов для Определение Минимального Количества Контрактов <<<<<<<<<<<<<<<<<<<<<<<<<<<<<<
# '''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''
path_startswith     = "CRYPTO"
Lot                 = 0.01      # Минимальное количество контрактов для которого производится расчёт
new_contract_size   = 1         # Размер контракта, с которого начинается расчёт
new_spread_cost     = 0         # Стоимость спреда, которая не должна быть превышена при размере позиции [Lot]
new_min_lot_pips_cost   = 0.01
```

```python
# Определение Минимального Количества Контрактов для заданного объёма <<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<
# '''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''
symbols_contracts_df = pd.DataFrame()                                  # Создаём пустой DataFrame для хранения информации о размерах  контрактов торговых инструментов

list_col_name = ['Symbol',"spread",
                 'ContractSize', 'new_contract_size_fo_iter',
                 "min_lot_pips_cost", "new_min_lot_pips_cost_fo_iter",
                 'min_lot_spread_cost', 'new_spread_cost_fo_iter']
add_list_col_name = True
new_position = 5
new_position_step = 2
customers_accounts_crm_mt5_df = imported["move_column"](symbols_contracts_df, list_col_name, add_list_col_name, new_position, new_position_step)

manager = imported['manager_connect_with_control']()
print(manager.SelectedAddAll())
print(manager.SelectedTotal())


for a in symbol_array:                                                 # Перебираем все элементы массива торговых инструментов                         
    if a.Path.startswith(path_startswith):
        
        contract_size   = a.ContractSize
        last_tick = manager.TickLastRaw(getattr(a, 'Symbol', None))    # Создаём объект TickLastRa
        spread                      = last_tick.ask - last_tick.bid
        if spread == 0: spread = a.Point  # Если спред равен нулю, то используем значение Point из объекта торгового инструмента
        min_lot_spread_cost         = spread * contract_size * Lot
        min_lot_pips_cost           = a.Point * contract_size * Lot

        new_contract_size_fo_iter   = 1#new_contract_size
        new_spread_cost_fo_iter     = 0#new_spread_cost
        new_min_lot_pips_cost_fo_iter  = 0#new_min_lot_pips_cost

        print(f"a_symbol = {a.Symbol}, new_contract_size = {new_contract_size}, new_spread_cost_fo_iter = {new_spread_cost_fo_iter}, new_min_lot_pips_cost_fo_iter = {new_min_lot_pips_cost_fo_iter}")
        # определяем минимальный кратный 10 или равный 1 размер контракта при котором стоимость спреда менее 0.1 <<<<<<<<<<<<<
        #if new_spread_cost_fo_iter < 0.1 and new_contract_size_fo_iter < 1000000 and new_min_lot_pips_cost_iter <= 0.01:

        print(f"a_symbol = {a.Symbol}")
        print(f"new_contract_size = {new_contract_size}, new_spread_cost = {new_spread_cost}")

        while (new_spread_cost_fo_iter < 0.1  or new_min_lot_pips_cost_fo_iter < 0.01):
            new_contract_size_fo_iter       = new_contract_size_fo_iter * 10
            new_spread_cost_fo_iter         = round(spread * new_contract_size_fo_iter * Lot, 2)   # Вычисляем стоимость спреда
            new_min_lot_pips_cost_fo_iter   = a.Point * new_contract_size_fo_iter * Lot
            print(f"new_contract_size = {new_contract_size_fo_iter}, new_spread_cost = {new_spread_cost_fo_iter}")

        new_contract_size_fo_iter   = new_contract_size_fo_iter                                              # Убираем последнее увиличение
        new_spread_cost_fo_iter     = new_spread_cost_fo_iter                                                # Убираем последнее увиличение
        new_min_lot_pips_cost_fo_iter = a.Point * new_contract_size_fo_iter * Lot

        """else: 
            new_contract_size_fo_iter = 0                                              
            new_spread_cost_fo_iter = 0 
            new_min_lot_pips_cost_fo_iter = 0"""

        print(f"a_symbol = {a.Symbol}, new_contract_size = {new_contract_size}, new_spread_cost_fo_iter = {new_spread_cost_fo_iter}")
        # >>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>

        # Добавляем данные в DataFrame
        symbols_contracts_df.loc[len(symbols_contracts_df)] = {
        'Symbol': a.Symbol, "spread": round(spread, 5),
        'ContractSize': contract_size, 'new_contract_size_fo_iter': int(new_contract_size_fo_iter),
        'min_lot_spread_cost': round(min_lot_spread_cost, 5), 'new_spread_cost_fo_iter': round(new_spread_cost_fo_iter, 5),
        "min_lot_pips_cost": round(min_lot_pips_cost, 5), "new_min_lot_pips_cost_fo_iter": round(new_min_lot_pips_cost_fo_iter, 5),
        }


        #a.ContractSize = int(new_contract_size)
        #print("ContractSize = ", a.ContractSize)
    else: pass
        #print(f"Пропускаем {a.Symbol} из-за несоответствия пути {a.Path} с {path_startswith}")

manager.SelectedDeleteAll()   
print("manager.Disconnect() = ", manager.Disconnect()) 

imported["pd_set_option"]("", symbols_contracts_df, 30)
```

```python
import pandas as pd

path_startswith = "FOREX"
manager = imported_functions['manager_connect']()
symbol_margin_base_urrency = pd.DataFrame(columns=['symbol', 'base', 'margin'])

# Print the selected objects in the manager
print(manager.SelectedAddAll())
print(manager.SelectedTotal())

for a in symbol_array:
    if a.Path.startswith(path_startswith):
        if a.CurrencyMargin != a.CurrencyBase:
            a.CurrencyMargin = a.CurrencyBase
            data = [[a.Symbol, a.CurrencyBase, a.CurrencyMargin]]                   # Data to be inserted into the DataFrame
            symbol_margin_base_urrency = pd.concat(                                 # Append the data to the DataFrame
                [symbol_margin_base_urrency, pd.DataFrame(data, columns=symbol_margin_base_urrency.columns)], 
                ignore_index=True
            )
display(symbol_margin_base_urrency)
```
