```python
# ВАРИАНТ №2: Загрузка данных SQL таблиц <<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<
# '''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''''
file_relative_path = "own_platform\\credits\\own_platforn_sql_main_main_01.txt"    # PROD server  Файл для подключения к БД и выполнения SQL запроса
#file_relative_path = "own_platform\\credits\\own_platforn_sql_main_stage_01.txt"  # STAGE server Файл для подключения к БД и выполнения SQL запроса
levels_up = 1

in_clause = ", ".join(f"'{item}'" for item in accounts_groups_list)                         # Формируем строку для SQL IN CLAUSE    
sql_tab = f"""SELECT * FROM `accountsGroups` WHERE name IN ({in_clause})""" # Запрос — только нужные строки!                                              
accounts_groups_df = imported["pd_read_sql"](sql_tab  + """;""", file_relative_path, levels_up)  #accounts_groups_sql_tab = "SELECT * FROM `accountsGroups`"   # таблица групп счетов
accounts_groups_id_list = accounts_groups_df['id'].tolist()


symbols_id_list = symbol_donor_df['symbolId'].tolist()                                               # Колонка → список (включая NaN)
symbols_id_str = imported['list_to_str'] (symbols_id_list, quotes=True)
print(f"Список symbolId для фильтрации: {symbols_id_str}")

#s_list = list(symbols_id_str ['name'].unique())

in_clause = ", ".join(f"'{item}'" for item in accounts_groups_id_list)
sql_tab  = f"SELECT * FROM `symbolsGroups` where symbolId in ({symbols_id_str}) and groupId in ({in_clause})"                                                       # таблица принадлежности символов группам символов
symbols_groups_donor_df = imported["pd_read_sql"](sql_tab  + """;""", file_relative_path, levels_up)
imported['pd_set_option']("[symbols_groups_donor_df] Данные по принадлежности символов группам символов", symbols_groups_donor_df, 50)
   
"""sql_tab = "SELECT * FROM symbolsMarkets"                                                      # соответствие символов и рынков
symbols_markets_df = imported["pd_read_sql"](sql_tab  + """;""", file_relative_path, levels_up)
imported['pd_set_option']("[symbols_markets_df] Данные по соответствию символов и рынков", symbols_markets_df, 5)"""

# отфильтровываем группы настройки которых транслируются на все торговые группы клиентов<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<
"""nativetradeengine_accounts_groups_df = accounts_groups_df[accounts_groups_df["name"].isin(accounts_groups_list)]    # Точное совпадение из списка
imported["pd_set_option"]("[nativetradeengine_accounts_groups_df] отфильтровываем группы настройки которых транслируются на все торговые группы клиентов", nativetradeengine_accounts_groups_df, 70)"""
```

```python
# создать ДФ настроек по клиентской группе берём symbols_groups_donor_df, и "размножаем" его под каждый новый символ.
# Берём symbols_groups_donor_df меняем symbolId на symbolId из symbols_fo_change_df, остальные данные оставляем неизменными.
# Мы получим ДФ с настройками групп символов для всех символов, для которых планируется изменение торговых условий.
# количество строк в итоговом ДF будет равняться произведением количесва строк в этих двух ДF
# ---------------------------------------------------------------------
# декартово произведение:
# каждая строка symbols_groups_donor_df
# копируется для каждого symbolId из symbols_fo_change_df
symbols_groups_result_df = (
    symbols_groups_donor_df.assign(_tmp=1)
    .merge(
        symbols_fo_change_df[['symbolId']].assign(_tmp=1),
        on='_tmp',
        suffixes=('', '_new')
    )
    .drop(columns=['_tmp', 'symbolId'])
    .rename(columns={'symbolId_new': 'symbolId'})
)
# >>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>

symbols_groups_result_df = symbols_groups_result_df[symbols_groups_donor_df.columns]    # опционально: порядок колонок как в исходном donor_df

"""symbols_groups_result_df['id'] = None                                                   # Заменить все значения baseCurrency на None
symbols_groups_result_df['createdAt'] = None
symbols_groups_result_df['updatedAt'] = None"""


imported['pd_set_option']("[symbols_groups_result_df] Данные по принадлежности символов группам символов", symbols_groups_result_df, 5)
```

```python
# К symbols_groups_result_df добавляем колонку с названием группы счетов (name переименовать в groupName), для которой эти настройки будут применены, по id группы из accounts_groups_df (колонку не переносить)
symbols_groups_result_df = symbols_groups_result_df.merge(
    accounts_groups_df[['id', 'name']],
    left_on='groupId',
    right_on='id',
    how='left',
    suffixes=('', '_group')
).drop(columns=['id_group']).rename(columns={'name': 'groupName'})

# Добавить одну колонку с именем символа (name переименовать в symbolName) взять из symbols_fo_change_df по symbolId, для удобства и наглядности
symbols_groups_result_df = symbols_groups_result_df.merge(
    symbols_fo_change_df[['symbolId', 'name']],
    left_on='symbolId',
    right_on='symbolId',
    how='left',
    suffixes=('', '_symbol')
).rename(columns={'name': 'symbolName'})#drop(columns=['symbolId']).

# Перенести groupName	symbolName в начало ДФ для удобства
cols = symbols_groups_result_df.columns.tolist()
cols = ['groupName', 'symbolName'] + [col for col in cols if col not in ['groupName', 'symbolName']]

cols = symbols_groups_result_df.columns.tolist()
cols = ['id', 'createdAt', 'updatedAt', 'groupName', 'symbolName'] + [col for col in cols if col not in ['id', 'createdAt', 'updatedAt', 'groupName', 'symbolName']]

symbols_groups_result_df = symbols_groups_result_df[cols]

imported["save_data_log_work_file"](symbols_groups_result_df, "symbols_groups_result_df.csv", directory_data_temp_files, directory_data_log_files)
imported['pd_set_option']("[symbols_groups_result_df] Данные по принадлежности символов группам символов", symbols_groups_result_df, 10)

```
